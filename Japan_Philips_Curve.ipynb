```python
import pandas as pd
import datetime
import matplotlib.pyplot as plt
import pandas_datareader as web
```

```python
start = datetime.datetime(1980, 1, 1)
end = datetime.datetime(2005, 12, 1)
```

```python
df_inf = web.DataReader('FPCPITOTLZGJPN', 'fred', start, end)
df_ur = web.DataReader('LRUNTTTTJPM156S', 'fred', start, end)
```

```python
df = pd.concat([df_ur, df_inf], axis=1).dropna()
df.columns = ['Unemployment Rate', 'Inflation Rate']
```

```python
plt.figure(figsize=(6, 9))
plt.scatter(-df['Unemployment Rate'], df['Inflation Rate'], s=200,
         color = 'red', alpha=0.6)
plt.title("A Map of Japan (1989-2019)", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Minus Unemployment Rate (-%)", fontsize=12)
plt.ylabel("Inflation Rate (CPI %)", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.5)
plt.figtext(0.15, 0.85, "Source: FRED(FPCPITOTLZGJPN, LRUNTTTTJPM156S)")

plt.show()
```
